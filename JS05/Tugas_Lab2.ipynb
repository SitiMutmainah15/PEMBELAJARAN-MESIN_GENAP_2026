```python
import pandas as pd

# Load data
df = pd.read_csv('spam.csv', encoding='latin-1')

# Ambil hanya kolom v1 dan v2
df = df.iloc[:, :2]

# Rename kolom
df = df.rename(columns={
    'v1': 'Labels',
    'v2': 'SMS'
})

# Encoding label
df['Labels'] = df['Labels'].map({
    'spam': 1,
    'ham': 0
})

df.head()
```

```python
X = df['SMS'].values
y = df['Labels'].values

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=50
)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# CountVectorizer dengan stop words
bow = CountVectorizer(stop_words='english')

# Fit hanya pada data training
X_train_bow = bow.fit_transform(X_train)

# Transform data testing
X_test_bow = bow.transform(X_test)

print("Jumlah fitur:", len(bow.get_feature_names_out()))
print("Dimensi data training:", X_train_bow.shape)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

mnb_bow = MultinomialNB()

# Training
mnb_bow.fit(X_train_bow, y_train)

# Prediksi
y_pred_bow = mnb_bow.predict(X_test_bow)

# Evaluasi
acc_bow = accuracy_score(y_test, y_pred_bow)

print("Akurasi CountVectorizer:", acc_bow)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_bow))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_bow))
```

CountVectorizer menghasilkan akurasi 98,30% dan recall spam 90%, sehingga mampu mengklasifikasikan spam dan ham dengan sangat baik.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# TF-IDF dengan stop words
tfidf = TfidfVectorizer(stop_words='english')

# Fit dan transform data training
X_train_tfidf = tfidf.fit_transform(X_train)

# Transform data testing
X_test_tfidf = tfidf.transform(X_test)

print("Jumlah fitur:", len(tfidf.get_feature_names_out()))
print("Dimensi data training:", X_train_tfidf.shape)
```

```python
# Model Multinomial Naive Bayes
mnb_tfidf = MultinomialNB()

# Training
mnb_tfidf.fit(X_train_tfidf, y_train)

# Prediksi
y_pred_tfidf = mnb_tfidf.predict(X_test_tfidf)

# Evaluasi
acc_tfidf = accuracy_score(y_test, y_pred_tfidf)

print("Akurasi TF-IDF:", acc_tfidf)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_tfidf))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_tfidf))
```

TF-IDF menghasilkan akurasi 96,05% dengan recall spam 73%, sehingga masih terdapat lebih banyak spam yang salah diklasifikasikan sebagai ham.

Pada percobaan ini, CountVectorizer merupakan fitur yang lebih optimal untuk klasifikasi spam.csv menggunakan Multinomial Naive Bayes.
